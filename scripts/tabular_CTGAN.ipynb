```python
import os
import time
import datetime
import pandas as pd
import numpy as np
import sys

from ctgan import CTGAN
```

```python
real_data = pd.read_csv('../data/simple_tabular_data.csv')
print(real_data)
```

```python
ctgan = CTGAN(epochs=10,verbose=True)
```

```python
# Names of the columns that are discrete
discrete_columns = [
    'workclass',
    'education',
    'marital-status',
    'occupation',
    'relationship',
    'race',
    'sex',
    'native-country',
    'income'
]

ctgan.fit(real_data, discrete_columns)
```

```python
synthetic_data = ctgan.sample(len(real_data))
```

```python
print(synthetic_data)
```

```python
synthetic_data.to_csv('../data/generated_ctgan_simple_tabular_data.csv', index=False)
```
