```python
import os
import time
import datetime
import pandas as pd
import numpy as np
import sys
from realtabformer import REaLTabFormer
```

```python

real_data = pd.read_csv('simple_tabular_data.csv')
print(real_data)
```

```python

numepochs=10
batchsize=10

transformer_model = REaLTabFormer(model_type="tabular",epochs=numepochs,batch_size=batchsize)


transformer_model.fit(real_data)
```

```python
synthetic_data = transformer_model.sample(len(real_data))
```

```python
print(synthetic_data)
```

```python
synthetic_data.to_csv('rtf_generated_simple_tabular_data.csv', index=True)
```
