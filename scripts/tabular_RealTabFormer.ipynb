```python
import os
import time
import datetime
import pandas as pd
import numpy as np
import sys
from realtabformer import REaLTabFormer
```

```python
real_data = pd.read_csv('../data/simple_tabular_data.csv')
print(real_data)
```

```python
numepochs=10
batchsize=10

transformer_model = REaLTabFormer(model_type="tabular",epochs=numepochs,batch_size=batchsize)

transformer_model.fit(real_data, n_critic=10)
```

```python
synthetic_data = transformer_model.sample(len(real_data))
```

```python
synthetic_data.to_csv('../data/generated_rtf_simple_tabular_data.csv', index=False)
```
